# Pokemon PSA Grader

Este proyecto consiste en una aplicación web y API que automatiza la clasificación del estado físico de las cartas Pokémon. Utiliza un modelo de **Deep Learning** para evaluar el desgaste (Condition Grading) y se integra con la **Pokémon TCG API** para proporcionar una valoración económica real y actualizada.

---

## **Plan de Trabajo**

### Fase 1: Base del proyecto y Mini-EDA
* **Setup:** Configuración de repo, entorno virtual y `requirements.txt`.
* **Exploración de PokeAPI:** Crear un servicio en Python que, dado un nombre de carta (ej. "Charizard"), devuelva su ID oficial y su precio de mercado base. Puedes añadir más información si ves necesario.
* **Recolección de Datos:** Buscar en Kaggle o recolectar imágenes de cartas en tres estados: `Perfecta`, `Normal` y `Dañada`.

### Fase 2: El Cerebro del Proyecto - IA
* **Preprocesamiento:** Redimensionar imágenes (IMPORTANTE) y normalizar píxeles.
* **Transfer Learning:** Entrenar un modelo CNN (Redes convolucionales)ligero MobileNetV2 utilizando los pesos de ImageNet para clasificar los 3 estados físicos.
* **Validación:** Asegurar que el modelo diferencia bien entre una carta nueva y una con bordes blancos (desgastada).
* **Exportación:** Guardar el modelo en formato `.pkl`.

### Fase 3: Construcción de la API
Implementación de los requisitos obligatorios:
* **Ruta `/health`:** Verificar estado del servicio.
* **Ruta `/card/{card_id}` (Path):** Devolver datos técnicos de la carta.
* **Ruta `/grade` (JSON Body/File):** Recibir la imagen, pasarla por el modelo y devolver la clasificación.
* **Ruta `/appraise` (Query Params):** Parámetro `currency=EUR` para personalizar la respuesta.

### Fase 4: Frontend y Despliegue
* **Interfaz Web:** Crear un Streamlit sencillo que permita:
    1. Buscar la carta.
    2. Subir una foto desde la cámara o galería.
    3. Mostrar el resultados.
* **Productivización:** Desplegar en **AWS EC2**(por tema de almacenamiento de RAM). Asegurar que las variables de entorno y las rutas públicas funcionan.

## **Tecnologías que usaremos.**

### Google Colab (Entrenamiento)
* **Por qué:** Proporciona acceso gratuito a GPUs (como la Tesla T4). Entrenar una red neuronal profunda en local sin hardware dedicado sería ineficiente.
* **Uso:** Será nuestro laboratorio para cargar el dataset de imágenes, aplicar **Transfer Learning** y exportar el modelo final en formato `.h5` o `.plk`.

### TensorFlow / Keras (MobileNetV2)
* **Por qué:** Es el estándar que hemos dominado en el bootcamp. Hemos elegido **MobileNetV2** porque es una arquitectura de Red Neuronal Convolucional (CNN) optimizada para dispositivos con pocos recursos.
* **Uso:** Actuará como el "ojo" del proyecto, clasificando las imágenes en estados (Mint, Played, Damaged) con un consumo de RAM mínimo, vital para el despliegue gratuito.

### FastAPI (El Motor)
* **Por qué:** Es el requisito principal del profesor (API REST). Lo preferimos sobre Flask por su velocidad y porque genera automáticamente la documentación interactiva Swagger(IMPORTANTE).
* **Uso:** Gestionará las rutas críticas: `/health` para estado, `/card` para búsqueda y `/grade` para la predicción de IA.

### Pokémon TCG API (Integración Externa)
* **Por qué:** Para que el proyecto destaque, no basta con clasificar una imagen; necesitamos datos reales. 
* **Uso:** FastAPI consultará esta API externa para obtener el precio de mercado base de la carta antes de aplicar el ajuste según el estado detectado por nuestra IA.

### Streamlit (La Cara Visual)
* **Por qué:** Permite crear interfaces de usuario profesionales usando solo Python. Nos da el factor "visual y de progresión".
* **Uso:** Servirá de puente para que el usuario suba la foto, la visualice y reciba la respuesta de nuestra API de AWS.

### AWS EC2 (Hosting del Backend)
* **Por qué:** El enunciado exige un despliegue accesible públicamente. Hemos elegido EC2 frente a Render porque nos ofrece **1GB de RAM** (frente a los 512MB de Render) y permite configurar memoria de intercambio (Swap).
* **Uso:** Alojará nuestra instancia permanente de FastAPI y el modelo de IA cargado en memoria.

### Streamlit Community Cloud (Hosting del Frontend)
* **Por qué:** Es la forma más rápida y limpia de desplegar la interfaz de Streamlit por separado.
* **Uso:** Mantendrá la URL pública que los usuarios usarán para interactuar con el sistema.
